```python
# Step 1: TensorFlow addition
import tensorflow as tf

a = tf.constant(5)
b = tf.constant(10)
print("TensorFlow Version:", tf.__version__)
print("TensorFlow Addition:", (a + b).numpy())
```

```python
# Step 2: Keras model demonstration
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

model = Sequential([
    Input(shape=(2,)),
    Dense(4, activation="relu"),
    Dense(1)
])
print("Keras Model:")
model.summary()
```

```python
# Step 3: Theano addition (legacy package; optional)
try:
    import theano
    import theano.tensor as T
    x = T.dscalar("x")
    y = T.dscalar("y")
    add = theano.function([x, y], x + y)
    print("Theano Addition:", add(5, 10))
except Exception:
    print("Theano is not supported or installed in this environment.")
```

```python
# Step 4: PyTorch addition
import torch

x = torch.tensor(5)
y = torch.tensor(10)
print("PyTorch Version:", torch.__version__)
print("PyTorch Addition:", (x + y).item())
```
